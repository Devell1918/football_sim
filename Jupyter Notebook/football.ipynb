```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option('display.max_columns', None)
df = pd.read_csv("cfb23.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
#create wins column and change 3rd Percent to numeric






#add yards/attempt allowed

df['3rd Percent'] = pd.to_numeric(df['3rd Percent'], errors='coerce')
df['Avg Turnover Margin per Game'] = pd.to_numeric(df['Avg Turnover Margin per Game'], errors='coerce')
df['Pass Yards/Attempt'] = pd.to_numeric(df['Pass Yards/Attempt'], errors='coerce')

df['Yards/Rush'] = pd.to_numeric(df['Yards/Rush'], errors='coerce')
df['Yds/Rush Allowed'] = pd.to_numeric(df['Yds/Rush Allowed'], errors='coerce')
df['Yards/Attempt Allowed'] = pd.to_numeric(df['Yards/Attempt Allowed'], errors='coerce')


df['Wins'] = pd.to_numeric(df['Win-Loss'].str.split('-').str[0], errors='coerce')
df_sorted_by_wins = df.sort_values(by='Wins', ascending=True)
df_sorted_by_wins.head()
```

```python
sns.lmplot(x="Wins", y="3rd Percent", data = df_sorted_by_wins, scatter_kws={'alpha': 0.3})
sns.lmplot(x="Wins", y="Avg Turnover Margin per Game", data = df_sorted_by_wins, scatter_kws={'alpha': 0.3})
sns.lmplot(x="Wins", y="Pass Yards/Attempt", data = df_sorted_by_wins, scatter_kws={'alpha': 0.3})
sns.lmplot(x="Wins", y="Yards/Rush", data = df_sorted_by_wins, scatter_kws={'alpha': 0.3})
sns.lmplot(x="Wins", y="Yds/Rush Allowed", data = df_sorted_by_wins, scatter_kws={'alpha': 0.3})
sns.lmplot(x="Wins", y="Yards/Attempt Allowed", data = df_sorted_by_wins, scatter_kws={'alpha': 0.3})
```
